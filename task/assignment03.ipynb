## 1. Feature Engineering

Feature engineering transforms raw incident attributes into predictive features optimized for machine learning. This section creates three categories of features based on Assignment 2's EDA findings:
1. Operational Complexity Features: Composite metrics measuring incident handling difficulty
2. Severity Features: Binary indicators addressing multicollinearity in severity classifications

### 1.1 Operational Complexity Features

Domain knowledge from Assignment 2's EDA is applied to engineer operational complexity features that capture incident handling difficulty. These features transform Assignment 2's correlation findings (sys_mod_count r=-0.368, reassignment_count r=-0.248) into actionable risk indicators. The engineered metrics quantify how many times an incident requires modification or reassignment—operational churn that directly correlates with SLA breach risk.

```python
# ===========================
# OPERATIONAL COMPLEXITY FEATURES
# ===========================
# Evidence from Assignment 2 EDA:
#   - sys_mod_count (r=-0.368) strongest SLA predictor
#   - reassignment_count (r=-0.248) second strongest
#   - sys_mod_count threshold at 3-4: >99% SLA (0-2 mods) vs <95% SLA (4+ mods)
#   - reassignment_count threshold at 0: 92.6% SLA (0 reassign) vs 84.7% SLA (1+ reassign)
#   - 7.9 percentage point performance gap from first reassignment

print("\n" + "=" * 80)
print("FEATURE ENGINEERING - OPERATIONAL COMPLEXITY")
print("=" * 80)

df_fe = df.copy()

# Activity Score: Weighted complexity metric
# Rationale for weights:
#   - System modifications: 1× (baseline, r=-0.368)
#   - Reassignments: 2× (r=-0.248, 7.9%p first-step impact)
#   - Weight justification:
#     * Step-function behavior (critical first occurrence)
#     * Conservative choice (avoids arbitrary high weights)
df_fe['activity_score'] = (
    df_fe['sys_mod_count'] +              # Troubleshooting iterations (1x weight)
    df_fe['reassignment_count'] * 2       # Routing failures (2x weight - higher SLA impact)
)

print("\nCreated Composite Metrics:")
print(f"activity_score = sys_mod_count + (reassignment_count × 2)")
print(f"Mean: {df_fe['activity_score'].mean():.2f}")
print(f"Median: {df_fe['activity_score'].median():.0f}")
print(f"Max: {df_fe['activity_score'].max():.0f}")

# Binary complexity flags based on EDA-identified thresholds
# Threshold justification:
#   - has_reassignment (>0): 7.9% SLA drop from first reassignment
#   - high_modification (>3): Performance cliff at 4+ modifications
#   - is_complex: Combined flag for incidents requiring extensive handling
df_fe['has_reassignment'] = (
    df_fe['reassignment_count'] > 0  # Any reassignment = routing failure
).astype(int)

df_fe['high_modification'] = (
    df_fe['sys_mod_count'] > 3  # >3 mods = performance cliff (EDA: 99% → <95%)
).astype(int)

df_fe['is_complex'] = (
    (df_fe['sys_mod_count'] > 4) |  # Extensive troubleshooting OR
    (df_fe['reassignment_count'] > 2)  # Multiple routing failures
).astype(int)

print("\nCreated Binary Flags:")
print(
    f"has_reassignment (>0): {df_fe['has_reassignment'].sum():,} ({df_fe['has_reassignment'].mean():.1%})"
)
print(
    f"high_modification (>3): {df_fe['high_modification'].sum():,} ({df_fe['high_modification'].mean():.1%})"
)
print(
    f"is_complex (>4 mods OR >2 reassign): {df_fe['is_complex'].sum():,} ({df_fe['is_complex'].mean():.1%})"
)

# ===========================
# VALIDATION: VERIFY FEATURE EFFECTIVENESS
# ===========================
# Purpose: Confirm engineered features capture real SLA performance differences
# Method: Calculate SLA compliance rate by feature value

# is_complex validation
print("\nFeature Validation:")
complex_sla = df_fe.groupby('is_complex')['made_sla'].mean()
print(f"Simple incidents (0): {complex_sla[0]:.1%} SLA compliance")
print(f"Complex incidents (1): {complex_sla[1]:.1%} SLA compliance")
print(
    f"Performance gap: {(complex_sla[0]-complex_sla[1])*100:.1f} percentage points"
)

print("\n" + "-" * 80)
print("Summary: Created 4 operational complexity features")
print("- activity_score (continuous)")
print("- has_reassignment, high_modification, is_complex (binary)")
print("-" * 80)
```

- The complexity flag shows strong discriminative power, with simple incidents achieving 98.3% SLA compliance compared to 68.9% for complex incidents (29.4 percentage point gap).
- The activity score captures compounding effects through 2× weighting of reassignments, reflecting Assignment 2's finding that reassignments have disproportionate SLA impact (7.9% drop from first reassignment: 92.6% → 84.7%)
- Threshold-based flags enable early intervention when incidents cross critical thresholds (first reassignment causing 7.9%p drop, or 4+ modifications marking the performance cliff), allowing proactive response before incidents escalate into full complexity.
- These engineered features translate Assignment 2's correlation analysis (sys_mod_count r=-0.368, reassignment_count r=-0.248) into operationally actionable metrics suitable for real-time scoring.

### 1.2 Severity Features

Assignment 2's EDA revealed multicollinearity among severity dimensions (priority, impact, urgency with r=0.75-0.89 correlations). To address this, severity classifications are converted into binary risk indicators that capture the critical high-vs-low severity distinction without redundant information. This transformation enables linear models to process severity information while preserving the substantial SLA performance gaps (38.8% for composite flag, 42.2-48.1% for individual dimensions) identified in Assignment 2.

```python
# ===========================
# SEVERITY FEATURES
# ===========================
# Purpose: Convert categorical severity levels to binary risk indicators
# Method: Binary encoding + composite severity flag
# Rationale:
#   - Priority, Impact, and Urgency exhibit strong multicollinearity (r=0.75-0.89)
#   - Using all three introduces redundancy in linear models
#   - Binary "high vs low" flags avoid multicollinearity while preserving signal
# Evidence from Assignment 2 EDA:
#   - Priority: Critical 53.3% vs Low 95.5% SLA (42.2% gap)
#   - Impact: High 51.6% vs Low 95.8% SLA (44.2% gap)
#   - Urgency: High 48.1% vs Low 96.2% SLA (48.1% gap)

print("\n" + "=" * 80)
print("FEATURE ENGINEERING - SEVERITY INDICATORS")
print("=" * 80)

# Binary severity flags
# Mapping logic:
#   - "1 - Critical" / "2 - High" → High severity (1)
#   - "3 - Moderate" / "4 - Low" → Low severity (0)
df_fe['is_high_priority'] = (
    df_fe['priority'].isin(['1 - Critical',
                            '2 - High'])  # Critical/High = 1, else 0
).astype(int)

df_fe['is_high_impact'] = (
    df_fe['impact'].isin(['1 - High'])  # High = 1, else 0
).astype(int)

df_fe['is_high_urgency'] = (
    df_fe['urgency'].isin(['1 - High'])  # High = 1, else 0
).astype(int)

print("\nCreated Binary Severity Flags:")
print(
    f"is_high_priority: {df_fe['is_high_priority'].sum():,} ({df_fe['is_high_priority'].mean():.1%})"
)
print(
    f"is_high_impact: {df_fe['is_high_impact'].sum():,} ({df_fe['is_high_impact'].mean():.1%})"
)
print(
    f"is_high_urgency: {df_fe['is_high_urgency'].sum():,} ({df_fe['is_high_urgency'].mean():.1%})"
)

# Composite severity flag
# Logic: Mark as high severity if ANY dimension (priority/impact/urgency) is high
# Rationale: Conservative approach captures all potentially high-risk incidents
df_fe['is_high_severity'] = (
    (df_fe['is_high_priority'] == 1) |  # High priority OR
    (df_fe['is_high_impact'] == 1) |  # High impact OR
    (df_fe['is_high_urgency'] == 1)  # High urgency = High severity
).astype(int)

print(f"\nCreated Composite Flag:")
print(
    f"is_high_severity (ANY dimension high): {df_fe['is_high_severity'].sum():,} ({df_fe['is_high_severity'].mean():.1%})"
)

# ===========================
# VALIDATION: VERIFY SEVERITY IMPACT
# ===========================
# Purpose: Confirm severity flags predict SLA performance
# Feature Validation
print("\nFeature Validation:")
sev_sla = df_fe.groupby('is_high_severity')['made_sla'].mean()
print(f"Low severity (0): {sev_sla[0]:.1%} SLA compliance")
print(f"High severity (1): {sev_sla[1]:.1%} SLA compliance")
print(f"Performance gap: {(sev_sla[0]-sev_sla[1])*100:.1f} percentage points")

print("\n" + "-" * 80)
print("Summary: Created 4 severity features")
print("- is_high_priority, is_high_impact, is_high_urgency (individual)")
print("- is_high_severity (composite OR logic)")
print("-" * 80)
```

- The is_high_severity flag achieves a 38.8 percentage point gap (87.2% vs 48.4%), 
comparable to the individual gaps from Assignment 2 (42.2% Priority, 44.2% Impact, 
48.1% Urgency). This confirms the composite OR-logic approach preserves substantial 
signal strength while eliminating multicollinearity—capturing high-risk incidents 
flagged by any severity dimension rather than requiring all three simultaneously.
- By converting the highly correlated severity variables (r=0.75-0.89) into binary flags, the features become independent risk indicators suitable for linear models while retaining the critical high-vs-low severity distinction that drives SLA outcomes.
- The composite flag uses OR logic rather than AND, ensuring all potentially at-risk incidents receive appropriate attention. This approach prioritizes recall over precision in the initial flagging stage.

### 1.3 Interaction Features

Multiplicative interaction terms capture compounding risk effects where combined factors create greater SLA breach probability than individual contributions alone. These features address Assignment 2's multivariate analysis findings that high-severity incidents with operational complexity face exponentially higher failure rates. Interaction terms enable models to identify cases where multiple risk factors converge—the highest-priority incidents for proactive intervention.

```python
# ===========================
# INTERACTION FEATURES
# ===========================
# Purpose: Capture multiplicative risk effects from combined factors
# Evidence from Assignment 2 Section 3.7 (Multivariate Analysis):
#   - Priority × Reassignment: 51.6%p compounding gap
#     Low-priority 0-reassign (97.7%) vs High-priority 0-reassign (46.1%)
#   - Knowledge × Priority: Variable effectiveness by priority
#     Moderate: +4.6% | Low: -9.1% | High/Critical: negative
#   - Group × Category: 60%p systematic gap
#     Best (100%) vs Worst (40%)

# Evidence from Assignment 2 Section 3.8 (Statistical Tests):
#   - Priority: χ² = 275.76 (strongest severity indicator)
#   - Priority_confirmation: χ² = 178.70 (complexity marker)
#   - Sys_mod: t = -32.44, mean diff 9.51 (194% more in breach cases)
#   - Reassignment: t = -20.98, mean diff 1.13 (147% more in breach cases)

print("\n" + "=" * 80)
print("FEATURE ENGINEERING - INTERACTION TERMS")
print("=" * 80)

# Severity × Complexity interaction
# Captures: Urgent incidents requiring extensive troubleshooting
# Hypothesis: Combined effect worse than sum of individual effects
df_fe['severity_complexity'] = (
    df_fe['is_high_severity'] *  # High severity AND
    df_fe['is_complex']
    # Captures 51.6%p compounding from Assignment 2 3.7
).astype(int)

# Priority Confirmation × Complexity interaction
# Captures: Procedurally flagged incidents that also require complex handling
# Note: u_priority_confirmation available at incident time (not post-incident)
# Hypothesis: Confirmed complex cases represent highest-risk category
df_fe['confirmed_complex'] = (
    df_fe['u_priority_confirmation']
    *  # Priority confirmed (procedural flag) AND
    df_fe['is_complex']  # Complex handling = Confirmed difficult case
).astype(int)

# High Priority × Reassignment interaction
# Captures: Urgent cases experiencing routing failures
# Hypothesis: Time-sensitive incidents lose critical time during handoffs
df_fe['priority_reassign'] = (
    df_fe['is_high_priority'] *  # High priority (time-sensitive) AND
    df_fe['has_reassignment']
    # Time-sensitive incidents + routing failure
    # Assignment 2 showed High priority drops from 46.1% → 36.7% with reassignment
).astype(int)

print("\nCreated Interaction Features:")
print(
    f"severity_complexity: {df_fe['severity_complexity'].sum():,} ({df_fe['severity_complexity'].mean():.1%})"
)
print(
    f"confirmed_complex: {df_fe['confirmed_complex'].sum():,} ({df_fe['confirmed_complex'].mean():.1%})"
)
print(
    f"priority_reassign: {df_fe['priority_reassign'].sum():,} ({df_fe['priority_reassign'].mean():.1%})"
)

# ===========================
# VALIDATION: VERIFY INTERACTION IMPACT
# ===========================
# Purpose: Confirm interaction terms capture additional risk beyond individual features
# Feature Validation
print("\nFeature Validation:")
risk_sla = df_fe.groupby('severity_complexity')['made_sla'].mean()
print(f"No risk (0): {risk_sla[0]:.1%} SLA compliance")
print(f"High risk (1): {risk_sla[1]:.1%} SLA compliance")
print(f"Gap: {(risk_sla[0]-risk_sla[1])*100:.1f} percentage points")

# Compare to individual feature gaps
individual_severity_gap = (sev_sla[0] - sev_sla[1]) * 100
individual_complexity_gap = (complex_sla[0] - complex_sla[1]) * 100
interaction_gap = (risk_sla[0] - risk_sla[1]) * 100

print(f"\nAdditive Hypothesis Test:")
print(f"Severity gap alone: {individual_severity_gap:.1f}%")
print(f"Complexity gap alone: {individual_complexity_gap:.1f}%")
print(
    f"Expected additive gap: {individual_severity_gap + individual_complexity_gap:.1f}%"
)
print(f"Actual interaction gap: {interaction_gap:.1f}%")

print("\n" + "-" * 80)
print(f"Feature Engineering Summary:")
print(f"Original features: {df.shape[1]}")
print(f"Engineered features: {df_fe.shape[1]}")
print(f"Features added: {df_fe.shape[1] - df.shape[1]}")
print("-" * 80)
```

- The severity_complexity interaction shows a 41.9 percentage point gap (86.9% vs 45.0%), 
stronger than either severity alone (38.8%p) or complexity alone (29.4%p). However, 
the combined effect (41.9%p) is substantially less than the expected additive effect 
(68.2%p), revealing significant overlap—many high-severity incidents are also complex. 
This sub-additive pattern (41.9% vs 68.2% expected) indicates the interaction term 
identifies the joint high-risk population rather than creating additional multiplicative 
effects. The 26.3 percentage point gap from additivity suggests approximately 38% 
shared variance [(68.2-41.9)/68.2] between severity and complexity dimensions.
- The confirmed_complex interaction identifies cases where procedural validation flags combine with operational complexity, representing the highest-risk category. This feature enables models to prioritize incidents flagged by both human judgment and system metrics.
- The priority_reassign interaction captures time-sensitive incidents losing critical time during routing failures. Given Assignment 2's finding that zero-reassignment incidents achieve 92.6% SLA while incidents with reassignment drop to 84.7% (7.9 percentage point gap), combining this with high priority creates a feature specifically targeting time-sensitive routing failures.
- Feature count increased from 32 to 43 (11 additions), remaining conservative to avoid curse of dimensionality. Each added feature addresses specific EDA findings: complexity (4 features), multicollinearity resolution (4 features), and interaction effects (3 features).

### 1.4 Feature Encoding

Feature encoding converts categorical and ordinal variables into numerical representations suitable for machine learning algorithms. With 43 features after engineering (from original 32), proper encoding is critical to ensure models can process different data types while preserving information. Three encoding strategies are employed based on variable characteristics: ordinal encoding for severity hierarchies (maintaining natural order), label encoding for high-cardinality categoricals (avoiding 100+ sparse features), and binary encoding for flags (ensuring 0/1 consistency).

#### 1.4.1 Ordinal Encoding

```python
# ===========================
# ORDINAL ENCODING
# ===========================
# Purpose: Preserve natural ordering in severity classifications
# Method: Map categorical labels to integers reflecting severity hierarchy
# Rationale:
#   - Priority, Impact, Urgency have natural low→high ordering
#   - Ordinal encoding preserves this ordering (unlike one-hot encoding)
#   - Enables models to learn monotonic relationships with SLA

print("\n" + "=" * 80)
print("FEATURE ENCODING - ORDINAL VARIABLES")
print("=" * 80)

df_enc = df_fe.copy()

# Priority encoding (1=Low → 4=Critical)
# Mapping: '4 - Low' → 1, '3 - Moderate' → 2, '2 - High' → 3, '1 - Critical' → 4
# Rationale: Higher numbers = higher severity = worse SLA performance
priority_map = {
    '4 - Low': 1,  # Lowest severity → smallest number
    '3 - Moderate': 2,  # Moderate severity
    '2 - High': 3,  # High severity
    '1 - Critical': 4  # Highest severity → largest number
}

df_enc['priority_enc'] = df_enc['priority'].map(priority_map)  # Apply ordinal mapping

# Impact encoding (1=Low → 3=High)
impact_map = {
    '3 - Low': 1,  # Low impact → smallest number
    '2 - Medium': 2,  # Medium impact
    '1 - High': 3  # High impact → largest number
}
df_enc['impact_enc'] = df_enc['impact'].map(
    impact_map)  # Apply ordinal mapping

# Urgency encoding (1=Low → 3=High)
urgency_map = {
    '3 - Low': 1,  # Low urgency → smallest number
    '2 - Medium': 2,  # Medium urgency
    '1 - High': 3  # High urgency → largest number
}
df_enc['urgency_enc'] = df_enc['urgency'].map(
    urgency_map)  # Apply ordinal mapping

print("\nOrdinal Encoded Variables:")
print(f"priority_enc: 1 (Low) → 4 (Critical)")
print(f"impact_enc: 1 (Low) → 3 (High)")
print(f"urgency_enc: 1 (Low) → 3 (High)")
print("Successfully encoded 3 ordinal variables")
```

#### 1.4.2 Label Encoding

```python
# ===========================
# LABEL ENCODING
# ===========================
# Purpose: Convert high-cardinality categorical variables to integers
# Method: LabelEncoder assigns unique integer to each category
# Rationale:
#   - assignment_group and category have many unique values (>50)
#   - No natural ordering exists
#   - Label encoding enables use in tree-based models
#   - Alternative (one-hot) would create 100+ sparse features
# Note: Ordinal relationship is arbitrary but acceptable for tree-based models

print("\n" + "=" * 80)
print("FEATURE ENCODING - CATEGORICAL VARIABLES")
print("=" * 80)

# assignment_group encoding
# From Assignment 2 EDA: 47.8% performance variance across groups (52.2%-100%)
# Strategy: Label encode to preserve information while avoiding dimensionality explosion
le_group = LabelEncoder()
df_enc['assignment_group'] = df_enc['assignment_group'].fillna('Unknown')
df_enc['assignment_group_enc'] = le_group.fit_transform(
    df_enc['assignment_group'])

# category encoding
# From Assignment 2 EDA: 26.4% performance gap across categories (67.9%-94.3%)
le_cat = LabelEncoder()
df_enc['category'] = df_enc['category'].fillna('Unknown')
df_enc['category_enc'] = le_cat.fit_transform(df_enc['category'])

print("\nLabel Encoded Variables:")
print(f"assignment_group_enc: {len(le_group.classes_)} unique groups")
print(f"category_enc: {len(le_cat.classes_)} unique categories")
print("Successfully encoded 2 categorical variables")
```

#### 1.4.3 Binary Encoding

```python
# ===========================
# BINARY ENCODING VERIFICATION
# ===========================
# Purpose: Ensure all binary flags are integer type (0/1)
# Method: Explicit type casting to int
# Rationale:
#   - Some binary features created via boolean operations (.astype(int))
#   - Others are native boolean columns (knowledge, u_priority_confirmation)
#   - Models require consistent integer type for binary features

print("\n" + "=" * 80)
print("FEATURE ENCODING - BINARY VARIABLES")
print("=" * 80)

# List of all binary features
binary_vars = [
    'knowledge',  # Process: knowledge base consulted
    'u_priority_confirmation',  # Process: priority confirmed
    'made_sla',  # Target: SLA compliance
    'has_reassignment',  # Complexity: any reassignment occurred
    'high_modification',  # Complexity: >3 system modifications
    'is_complex',  # Complexity: composite complexity flag
    'is_high_priority',  # Severity: priority is Critical/High
    'is_high_impact',  # Severity: impact is High
    'is_high_urgency',  # Severity: urgency is High
    'is_high_severity',  # Severity: composite severity flag
    'severity_complexity',  # Interaction: severity × complexity
    'confirmed_complex',  # Interaction: confirmation × complexity
    'priority_reassign'  # Interaction: priority × reassignment
]

# Convert all to integer type
for var in binary_vars:
    if var in df_enc.columns:
        df_enc[var] = df_enc[var].astype(int)

verified_count = len([v for v in binary_vars if v in df_enc.columns])
print(f"\nBinary Variable Verification: {verified_count} variables")
print("All binary variables verified as integer type (0/1)")
```

- Ordinal encoding preserves severity hierarchy by mapping Priority (1-4), Impact (1-3), and Urgency (1-3) to ascending integers, maintaining the natural low→high ordering identified in Assignment 2. This enables linear models to learn monotonic relationships where each severity increase corresponds to proportional SLA decline.
- Label encoding handles high cardinality efficiently with 65 assignment groups (47.8% performance variance: 52.2%-100.0%) and 47 categories (26.4% gap: 67.9%-94.3%), avoiding creation of 112 sparse one-hot encoded features while preserving group-level information. Tree-based models can effectively split on these encoded integers to identify high/low performing groups without requiring ordinal relationships.
- Binary encoding ensures model compatibility by explicitly converting all 12 binary candidate features (target made_sla excluded) to integer type, preventing type mismatch errors during model training. This standardization is critical for algorithms expecting 
numerical inputs and enables proper interpretation of coefficients and feature importance metrics.
- 18 features require encoding (3 ordinal + 2 label + 13 binary flags including target), with 12 binary features used as candidates (made_sla excluded as target).

## 2. Feature Selection

Feature selection identifies the most informative predictors and reduces model 
complexity. A total of 20 candidate features are available for model training: 
6 numerical (3 ordinal-encoded severity + 2 raw operational metrics + 1 engineered 
activity score), 2 categorical (label-encoded assignment group and category), 
and 12 binary flags (2 original process indicators + 10 engineered complexity/
severity/interaction flags). Selecting the most predictive features improves 
model performance, training efficiency, and interpretability by removing irrelevant 
or redundant features. This is particularly important given 6,729 incidents—
maintaining a healthy sample-to-feature ratio prevents overfitting while retaining 
predictive power.

Two complementary methods are employed to ensure comprehensive feature coverage:

1. Filter Method (ANOVA F-test): Statistical test measuring univariate feature-target relationships. This model-agnostic approach evaluates each feature independently based on statistical significance, providing a simple yet effective way to filter out less important features.

2. Embedded Method (Random Forest): Model-based importance capturing feature interactions and non-linear relationships. This method evaluates features within the context of ensemble decision trees, identifying which features contribute most to prediction accuracy.

The union of both methods ensures we capture both statistical significance (ANOVA) and model-specific relevance (Random Forest), providing a robust feature set that balances theoretical rigor with practical performance. This approach validates Assignment 2's key findings—features like `sys_mod_count` (r=-0.368, Assignment 2's strongest predictor) and `assignment_group_enc` (47.8% SLA variance) appear in both selections, confirming their importance from multiple perspectives.

### 2.1 Define Candidate Features

### 2.8 Class Imbalance Handling

Class weight balancing addresses the 6:1 SLA Met:Breach imbalance without synthetic sampling. The chosen approach (class_weight='balanced') assigns higher penalty to minority class misclassification, mathematically compensating for imbalance by making each breach error ~6.1× higher loss weight than a false alarm. This forces models to prioritize learning breach patterns despite having only 758 breach training examples.

```python
# ===========================
# CLASS IMBALANCE HANDLING
# ===========================
# Purpose: Address 6:1 SLA Met:Breach imbalance without synthetic sampling
# Method: class_weight='balanced' in model training
# Rationale:
#   - SMOTE not used due to overfitting risk (see detailed analysis below)
#   - Class weights assign higher penalty to minority class misclassification
#   - Training on real data only ensures production reliability
#   - Weight formula: n_samples / (n_classes * np.bincount(y))

print("\n" + "=" * 80)
print("CLASS IMBALANCE HANDLING")
print("=" * 80)

X_train_final = X_train_scaled.copy()  # Final training features (no SMOTE)
y_train_final = y_train.copy()  # Final training target (no SMOTE)

print(f"\nChosen Approach: class_weight='balanced'")
print(f"Method: Assign inverse-frequency weights to classes")
print(f"Total samples: {len(X_train_final):,}")
print(
    f"Class 0 (Breach): {(y_train_final==0).sum():,} ({(y_train_final==0).mean():.1%})"
)
print(
    f"Class 1 (Met): {(y_train_final==1).sum():,} ({(y_train_final==1).mean():.1%})"
)
print(
    f"Imbalance ratio: {(y_train_final==1).sum() / (y_train_final==0).sum():.2f}:1"
)


# Calculate class weights
n_samples = len(y_train_final)  # Total training samples
n_classes = 2  # Binary classification
breach_weight = n_samples / (n_classes *
                             (y_train_final == 0).sum())  # Weight for class 0
met_weight = n_samples / (n_classes *
                          (y_train_final == 1).sum())  # Weight for class 1

print(f"\nCalculated Class Weights:")
print(f"Class 0 (Breach): {breach_weight:.2f}x (higher penalty)")
print(f"Class 1 (Met): {met_weight:.2f}x (lower penalty)")
print(
    f"Model penalizes missed breaches {breach_weight/met_weight:.1f}x more than false alarms"
)
print("Data remains in original 6:1 ratio (no synthetic sampling)")
```

Why SMOTE Was Not Used:
- SMOTE (Synthetic Minority Over-sampling Technique) was not used in this analysis. 
The training set's 758 breach samples, while representing the minority class in a 
6:1 imbalance, provide sufficient diversity for learning real-world failure patterns 
given the 58:1 samples-per-feature ratio (758 breach samples / 13 features), which 
far exceeds the recommended 10:1 minimum for reliable model training.
- Using class_weight='balanced' (applies to models that support class_weight, e.g., Logistic Regression and Random Forest) achieved effective generalization without artificial data, with Random Forest demonstrating an 8.4% train-test gap and 84.2% recall. This 
approach assigns 3.55× penalty weight to breach misclassifications (vs 0.58× for 
met-class), mathematically compensating for the 6:1 imbalance by assigning approximately 6.1× higher loss weight to breach errors than to false alarms, forcing the model to prioritize 
learning breach patterns.
- Training exclusively on real incidents ensures the model learns actual breach patterns 
without risk of overfitting to interpolated/artificial patterns that don't exist in 
production environments. The 758 breach samples span 65 different assignment groups, 
multiple severity levels, and wide operational metric ranges (sys_mod_count: 0-50+, 
reassignment_count: 0-10+), providing the diversity needed for generalizable learning 
as evidenced by stable cross-validation performance (CV F1: 0.7244 ±0.0130) and 
successful test-set generalization.

Key Observations:
- Class weighting prevents minority class neglect by assigning 3.55× penalty weight to breach misclassifications (vs 0.58× for met-class), forcing models to prioritize learning breach patterns despite having only 758 breach training examples.
- With 5,383 training samples (758 breaches, 4,625 met-SLA), the dataset provides adequate volume for model learning.
- Training exclusively on real incidents ensures deployed predictions generalize to new incidents rather than overfitting to idealized patterns.

### 3.2 Overfitting and Underfitting

Model performance is evaluated using F1-Score for Class 0 (SLA Breach) rather 
than overall accuracy, as the 6:1 class imbalance allows a naive "always predict 
met-SLA" classifier to achieve 85.9% accuracy while detecting zero breaches. 
F1-Score's harmonic mean of precision and recall ensures both breach detection 
(recall) and prediction accuracy (precision) contribute equally to evaluation, 
preventing models from gaming the metric by predicting the majority class.

Model generalization is evaluated by comparing train F1, test F1, and 5-fold 
cross-validation F1. Performance gaps are classified as: Gap < 0 (underfitting), 
Gap 0-10% (good generalization), Gap 10-15% (moderate overfitting), Gap >15% 
(severe overfitting).

Cross-validation uses macro-averaged F1 to provide a class-balance-agnostic stability check, while the primary decision metric for reporting remains Class 0 (breach) F1.

```python
# ===========================
# OVERFITTING/UNDERFITTING EVALUATION
# ===========================
# Purpose: Detect generalization issues via train-test performance gap
# Method: Compare Train F1, Test F1, and 5-fold Cross-Validation F1
# Rationale:
#   - Gap < 0: Underfitting (model too simple)
#   - Gap 0-10%: Good generalization
#   - Gap 10-15%: Moderate overfitting
#   - Gap >15%: Severe overfitting
# Metric: F1-Score for Class 0 (SLA Breach) - primary business concern

print("\n" + "=" * 80)
print("OVERFITTING / UNDERFITTING ANALYSIS")
print("=" * 80)

results = []

for model_name, model in models.items():
    print(f"\n{model_name}")
    print("-" * 80)

    # Test predictions
    y_test_pred = model.predict(X_test_scaled)  # Predict on test set

    # Train predictions
    y_train_pred = model.predict(X_train_final)  # Predict on train set

    # ===========================
    # TEST METRICS
    # ===========================
    # Focus on Class 0 (Breach) - pos_label=0
    test_precision = precision_score(
        y_test,
        y_test_pred,
        pos_label=0,  # Focus on breach class
        zero_division=0  # Handle division by zero
    )
    test_recall = recall_score(
        y_test,
        y_test_pred,
        pos_label=0,  # Focus on breach class
        zero_division=0)
    test_f1 = f1_score(
        y_test,
        y_test_pred,
        pos_label=0,  # Focus on breach class
        zero_division=0)
    test_acc = accuracy_score(y_test, y_test_pred)  # Overall accuracy

    # ===========================
    # TRAIN METRICS
    # ===========================
    train_precision = precision_score(
        y_train_final,
        y_train_pred,
        pos_label=0,  # Focus on breach class
        zero_division=0)
    train_recall = recall_score(
        y_train_final,
        y_train_pred,
        pos_label=0,  # Focus on breach class
        zero_division=0)
    train_f1 = f1_score(
        y_train_final,
        y_train_pred,
        pos_label=0,  # Focus on breach class
        zero_division=0)

    # ===========================
    # CROSS-VALIDATION
    # ===========================
    # 5-fold stratified CV for stable performance estimate
    # Note: scoring='f1_macro' for CV stability across classes; reporting focuses on Class 0 (breach) F1.
    cv_scores = cross_val_score(
        model,
        X_train_final,  # Training features
        y_train_final,  # Training target
        cv=5,  # 5 folds
        scoring='f1_macro',  # Macro-averaged F1 (balanced)
        n_jobs=-1  # Use all CPU cores
    )

    # ===========================
    # GAP ANALYSIS
    # ===========================
    gap = train_f1 - test_f1  # Train-test gap
    gap_pct = (gap / train_f1 * 100) if train_f1 > 0 else 0  # Gap percentage

    # Store results
    results.append({
        'Model': model_name,
        'Test F1': test_f1,
        'Test Precision': test_precision,
        'Test Recall': test_recall,
        'Train F1': train_f1,
        'Train Precision': train_precision,
        'Train Recall': train_recall,
        'CV F1': cv_scores.mean(),
        'CV Std': cv_scores.std(),
        'Gap': gap,
        'Gap %': gap_pct
    })

    # Display metrics
    print(f"Test Metrics (Class 0 - SLA Breach):")
    print(
        f"  Precision: {test_precision:.4f}, Recall: {test_recall:.4f}, F1: {test_f1:.4f}"
    )
    print("Train Metrics:")
    print(
        f" Precision: {train_precision:.4f}, Recall: {train_recall:.4f}, F1: {train_f1:.4f}"
    )
    print(
        f"Cross-Validation: CV F1 = {cv_scores.mean():.4f} (±{cv_scores.std():.4f})"
    )
    print(f"Gap Analysis: Train F1 - Test F1 = {gap:.4f} ({gap_pct:.1f}%)")

    # Use gap_pct (percentage) for consistent interpretation
    if gap < 0:
        status = "UNDERFITTING"
    elif gap_pct < 10:
        status = "GOOD GENERALIZATION"
    elif gap_pct < 15:
        status = "MODERATE OVERFITTING"
    else:
        status = "SEVERE OVERFITTING"
    print(f"Status: {status}")

print(f"\nSUMMARY: {len(models)} models evaluated")
```

- Random Forest demonstrates production-ready performance with 8.4% train-test gap, placing it firmly in the "good generalization" zone (0-10% threshold). Cross-validation variance of ±0.0130 confirms stable performance across different data splits. The model successfully balances breach detection (84.2% recall) with reasonable precision (42.44%), providing a practical compromise for operational deployment where catching most breaches while tolerating some false alarms is preferable to missing critical incidents.
- Class weight balancing (3.55× breach penalty vs 0.58× met-SLA penalty) proved essential for handling the 6:1 imbalance without generating artificial data. This mathematical weighting compensated for having only 758 breach examples in training by assigning approximately 6.1× higher loss weight to breach errors than to met-SLA errors, forcing the model to prioritize learning breach patterns.
- Feature selection from 20 to 13 features prevented overfitting while maintaining predictive power. The final 414:1 sample-to-feature ratio (5,383 samples / 13 features) ensured sufficient data density for reliable model learning. This 35% dimensionality reduction eliminated redundant variables while retaining the most informative predictors identified through ANOVA F-test and Random Forest importance analysis.
- Logistic Regression's minimal underfitting (-0.2% gap) disguises its fatal flaw: catastrophically low precision (33.97%) means it generates false alarms for 66% of breach predictions. While achieving excellent recall (94.21%), the model is operationally unusable because it would overwhelm support teams with incorrect alerts. Its linear structure fundamentally cannot capture the non-linear relationships in operational metrics like sys_mod_count and reassignment_count interactions.
- Gradient Boosting's severe overfitting (19.3% gap, exceeding the 15% threshold) combines with unacceptably low recall (32.11%) to render it unsuitable for deployment. The model misses 68% of actual breaches—directly failing the primary business objective of proactive breach detection. Despite achieving higher precision (55.45%) than Random Forest, this advantage is meaningless when the model fails to identify most incidents requiring intervention. Sequential learning memorized training noise rather than learning generalizable breach patterns.

### 4.1 ROC & Precision-Recall Curves

```python
# ===========================
# ROC & PRECISION-RECALL CURVES
# ===========================
# Purpose: Evaluate model's discriminative power across decision thresholds
# Method: Plot ROC (True/False Positive Rate) and PR (Precision/Recall)
# Rationale:
#   - ROC AUC: Overall class separation ability (higher = better)
#   - PR AUC: Performance on imbalanced dataset (more informative than ROC)
#   - Baseline: Random classifier would achieve 0.5 ROC AUC, ~14.1% PR AUC

print("\n" + "=" * 80)
print("ROC & PRECISION-RECALL CURVES")
print("=" * 80)

# Get prediction probabilities
y_pred_proba_class1 = best_model.predict_proba(
    X_test_scaled_df)[:, 1]  # P(SLA Met)
y_pred_proba_class0 = best_model.predict_proba(
    X_test_scaled_df)[:, 0]  # P(SLA Breach)

# ===========================
# ROC CURVE
# ===========================
# Interpretation:
#   - X-axis: False Positive Rate (false alarms)
#   - Y-axis: True Positive Rate (recall)
#   - AUC: Area Under Curve (0.5=random, 1.0=perfect)

fpr, tpr, _ = roc_curve(
    y_test,  # True labels
    y_pred_proba_class1  # Predicted probabilities for class 1
)

roc_auc = auc(fpr, tpr)  # Calculate area under ROC curve

plt.figure(figsize=(8, 6))
plt.plot(fpr,
         tpr,
         color='darkorange',
         lw=2,
         label=f'ROC Curve (AUC = {roc_auc:.3f})')
plt.plot([0, 1], [0, 1],
         color='navy',
         lw=2,
         linestyle='--',
         label='Random Classifier (AUC = 0.500)')
plt.xlabel('False Positive Rate', fontsize=12, fontweight='bold')
plt.ylabel('True Positive Rate (Recall)', fontsize=12, fontweight='bold')
plt.title(f'ROC Curve - {best_model_name}', fontsize=14, fontweight='bold')
plt.legend(loc='lower right', fontsize=10)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print(f"\nROC-AUC Score: {roc_auc:.4f}")

# ===========================
# PRECISION-RECALL CURVE
# ===========================
# Interpretation:
#   - X-axis: Recall (how many breaches detected)
#   - Y-axis: Precision (how many predictions were correct)
#   - Baseline: Random classifier achieves ~14.1% (class proportion)
#   - More informative than ROC for imbalanced datasets

precision, recall, _ = precision_recall_curve(
    y_test,  # True labels
    y_pred_proba_class0,  # Predicted probabilities for class 0 (breach)
    pos_label=0  # Focus on breach class
)
ap_score = average_precision_score(
    y_test,  # True labels
    y_pred_proba_class0,  # Predicted probabilities for class 0
    pos_label=0  # Focus on breach class
)

plt.figure(figsize=(8, 6))
plt.plot(recall,
         precision,
         color='blue',
         lw=2,
         label=f'PR Curve (AP = {ap_score:.3f})')
baseline_ap = (y_test == 0).mean()  # Baseline = class proportion
plt.axhline(y=baseline_ap,
            color='red',
            linestyle='--',
            label=f'Baseline (Random): {baseline_ap:.3f}')
plt.xlabel('Recall (Breach Detection Rate)', fontsize=12, fontweight='bold')
plt.ylabel('Precision', fontsize=12, fontweight='bold')
plt.title(f'Precision-Recall Curve - {best_model_name}',
          fontsize=14,
          fontweight='bold')
plt.legend(loc='upper right', fontsize=10)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Average Precision: {ap_score:.4f}")
print(f"Improvement over Baseline: {(ap_score/baseline_ap - 1)*100:.1f}%")
```

An ROC-AUC of 0.8986 (~0.90) is near the “excellent discrimination” threshold (≥0.90), indicating robust separation between SLA breaches and compliances across decision thresholds. The Average Precision (AP) of 0.524 represents a 271.1% lift over the 14.1% random baseline, demonstrating effectiveness despite the 6:1 class imbalance. While ROC-AUC signals strong overall discrimination, the moderate AP (0.524 vs. 1.0) reflects the inherent trade-off of maintaining high precision while achieving ~84% breach recall in an imbalanced setting.

### 4.3 Permutation Importance

```python
# ===========================
# PERMUTATION IMPORTANCE
# ===========================
# Purpose: Measure feature importance by shuffling (model-agnostic)
# Method: Randomly permute each feature, measure F1-Score drop
# Rationale:
#   - More reliable than built-in importance (avoids split-based bias)
#   - Measures actual impact on predictions (not just splits)
#   - Evaluated on test set (reflects real-world importance)
#   - Model-agnostic (works for any algorithm)

print("\n" + "=" * 80)
print("PERMUTATION IMPORTANCE")
print("=" * 80)

perm_importance = permutation_importance(
    best_model,
    X_test_scaled,
    y_test,
    n_repeats=10,  # Repeat 10 times for stability
    random_state=42,
    scoring='f1_macro',  # Macro-averaged F1
    n_jobs=-1)

perm_imp_df = pd.DataFrame({
    'Feature': X_selected.columns,
    'Importance': perm_importance.importances_mean,
    'Std': perm_importance.importances_std
}).sort_values('Importance', ascending=False)

print(f"\nTop 10 Features (by F1-Score Drop):")
for rank, row in enumerate(perm_imp_df.head(10).itertuples(), 1):
    print(
        f"{rank:2d}. {row.Feature:30s} {row.Importance:.4f} (±{row.Std:.4f})"
    )

# Visualization
plt.figure(figsize=(10, 8))
top_10 = perm_imp_df.head(10)
plt.barh(range(len(top_10)),
         top_10['Importance'],
         xerr=top_10['Std'],
         color='coral',
         alpha=0.8,
         edgecolor='black')
plt.yticks(range(len(top_10)), top_10['Feature'])
plt.xlabel('Importance (F1-Score Drop when Permuted)',
           fontsize=12,
           fontweight='bold')
plt.title(f'Permutation Importance - {best_model_name}',
          fontsize=14,
          fontweight='bold')
plt.gca().invert_yaxis()
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

Confirmed_complex emerges as the second-most critical feature (0.0371), validating the complexity flag’s value. Notably, reassignment_count (0.0100) ranks #3 despite lower Gini importance (#8), suggesting its impact is underestimated by split-based metrics. The low standard deviations (±0.0019 to ±0.0155) indicate stable importance estimates across 10 permutation runs. Computed with f1_macro on the held-out test set for evaluation only (not used for tuning).

### 4.4 SHAP Analysis

```python
# ===========================
# SHAP ANALYSIS
# ===========================
# Purpose: Explain individual predictions using Shapley values
# Method: TreeExplainer for Random Forest (fast, exact for trees)
# Rationale:
#   - Shapley values: game-theoretic approach to feature attribution
#   - Each feature gets credit for its contribution to prediction
#   - Summary plot: global feature importance + value effects
#   - Waterfall plot: explain specific high-risk prediction

print("\n" + "=" * 80)
print("SHAP ANALYSIS (SHapley Additive exPlanations)")
print("=" * 80)

# Prepare data
X_train_final_df = pd.DataFrame(X_train_final, columns=X_selected.columns)

# Create TreeExplainer (fast for tree-based models)
explainer = shap.TreeExplainer(best_model)

# Calculate SHAP values
shap_values = explainer.shap_values(X_test_scaled_df)

# Handle SHAP values format (list for binary classification)
if isinstance(shap_values, list):
    shap_values_breach = shap_values[0]  # Class 0 (SLA Breach)
    print("Using class 0 (SLA Breach) SHAP values")
else:
    shap_values_breach = shap_values
    print("SHAP values computed")

# ===========================
# SHAP SUMMARY PLOT
# ===========================
# Interpretation:
#   - Y-axis: Features ranked by importance
#   - X-axis: SHAP value (impact on prediction)
#   - Color: Feature value (red=high, blue=low)
#   - Positive SHAP: pushes prediction toward breach
#   - Negative SHAP: pushes prediction toward met-SLA
plt.figure(figsize=(12, 8))
shap.summary_plot(shap_values_breach,
                  X_test_scaled_df,
                  plot_type="dot",
                  show=False)
plt.title(
    f'SHAP Summary Plot - {best_model_name}\nFeature Impact on SLA Breach Prediction',
    fontsize=14,
    fontweight='bold',
    pad=20)
plt.tight_layout()
plt.show()

# ===========================
# SHAP WATERFALL PLOT (High-Risk Prediction)
# ===========================
# Purpose: Explain a specific high-risk prediction in detail
# Method: Select incident with highest predicted breach probability

# Find highest-risk prediction
y_pred_proba = best_model.predict_proba(X_test_scaled_df)[:, 0]  # P(Breach)
high_risk_idx = np.argmax(y_pred_proba)

breach_prob = y_pred_proba[high_risk_idx]
actual_label = y_test.iloc[high_risk_idx] if hasattr(
    y_test, 'iloc') else y_test[high_risk_idx]
actual = "BREACH" if actual_label == 0 else "MET"

print(f"\nHighest Risk Prediction:")
print(f"Test Set Index: {high_risk_idx}")
print(f"Predicted Breach Probability: {breach_prob:.1%}")
print(f"Actual Outcome: SLA {actual}")

# Create waterfall plot
# Extract SHAP values for this prediction
if isinstance(shap_values, list):
    sample_shap_values = shap_values[0][high_risk_idx]
    if sample_shap_values.ndim == 2:
        sample_shap_values = sample_shap_values[:, 0]
    base_val = explainer.expected_value
    base_value = float(base_val[0]) if isinstance(
        base_val, (list, np.ndarray)) else float(base_val)
else:
    sample_shap_values = shap_values[high_risk_idx]
    if sample_shap_values.ndim == 2:
        sample_shap_values = sample_shap_values[:, 0]
    base_val = explainer.expected_value
    base_value = float(base_val[0]) if isinstance(
        base_val, (list, np.ndarray)) else float(base_val)

# Get feature values
sample_data = X_test_scaled_df.iloc[high_risk_idx].values
if sample_data.ndim != 1:
    sample_data = sample_data.flatten()

# Create Explanation object
explanation = shap.Explanation(
    values=sample_shap_values,
    base_values=base_value,
    data=sample_data,
    feature_names=X_test_scaled_df.columns.tolist())

# Generate waterfall plot
plt.figure(figsize=(10, 6))
shap.plots.waterfall(explanation, show=False)
plt.title(
    f'SHAP Waterfall - High-Risk Prediction\nBreach Probability: {breach_prob:.1%}',
    fontsize=13,
    fontweight='bold')
plt.tight_layout()
plt.show()
```

The SHAP summary plot identifies assignment_group_enc and activity_score as the top two features by mean absolute SHAP value, indicating their consistent influence across predictions. The violin plot shows clear red–blue separation for assignment_group_enc: higher encoded values (red) push predictions toward breach (positive SHAP), while lower values (blue) push toward met-SLA (negative SHAP). Note: assignment_group_enc is label-encoded; the code order is arbitrary, so “high code = high risk” reflects group-specific effects rather than ordinal meaning. For activity_score, the predominantly blue distribution on the negative side indicates that low activity strongly protects against breaches, whereas occasional red points on the positive side show that high activity contributes to breach risk.

For the highest-risk prediction (predicted breach, actual: BREACH), the waterfall decomposition of the model’s raw output shows how features shift the prediction from the model’s base value toward breach. Severity indicators make the largest positive contributions (e.g., is_high_priority +0.10, is_high_severity +0.09). Complexity flags (is_complex +0.04, confirmed_complex +0.04, severity_complexity +0.04, high_modification +0.03) add substantial additional positive impact, demonstrating compounding effects. Operational metrics (reassignment_count +0.03, has_reassignment +0.02, activity_score +0.02) further increase the prediction, with remaining features contributing smaller adjustments, resulting in a high-risk decision. Note: SHAP values here are on the model’s raw (tree path dependent) output, so we report contributions without percentage-point units. Computed on the held-out test set for evaluation only (not used for tuning).
